```python
import pandas as pd
df = pd.read_csv('../datasets/dps-thesis-dataset.csv')
df.info()
```

## Correlations

```python
df_numbers = df.select_dtypes(include=[np.number])
# remove spurrious columns
del df_numbers['participant_number']
del df_numbers['Difference_C2_C1']
del df_numbers['interaction_to_session_ratio']
del df_numbers['total_interaction_count']

corr_matrix = df_numbers.corr()
corr_matrix
```

## Correlation Matrix

```python
# seaborn plot of correlation matrix
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm', square=True)
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

## Bivariate Plots

```python
# seaborn bivate plot matrix for all features
sns.pairplot(df_numbers)
plt.suptitle('Bivariate Plot Matrix of Numerical Features', y=1.02)
plt.show()
```
